# Seaborn - Statistical Data Visualisation
Distribution plots, relational plots, categorical plots, pair grids, and themes.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

print('Seaborn version:', sns.__version__)
sns.set_theme(style='whitegrid', palette='muted')
```

## 1. Load Built-in Datasets

```python
tips     = sns.load_dataset('tips')
iris     = sns.load_dataset('iris')
penguins = sns.load_dataset('penguins').dropna()

print('tips shape:', tips.shape)
print('iris shape:', iris.shape)
print('penguins shape:', penguins.shape)
tips.head()
```

## 2. Distribution Plots

```python
fig, axes = plt.subplots(1, 3, figsize=(13, 4))

sns.histplot(tips['total_bill'], bins=20, kde=True, ax=axes[0], color='steelblue')
axes[0].set_title('Histogram + KDE')

sns.kdeplot(data=tips, x='total_bill', hue='sex', fill=True, alpha=0.4, ax=axes[1])
axes[1].set_title('KDE by Sex')

sns.ecdfplot(data=tips, x='total_bill', hue='time', ax=axes[2])
axes[2].set_title('ECDF by Meal Time')

plt.tight_layout()
plt.show()
```

## 3. Relational Plots

```python
g = sns.relplot(
    data=tips,
    x='total_bill', y='tip',
    hue='sex', style='smoker', size='size',
    sizes=(30, 200), alpha=0.7,
    height=5, aspect=1.4
)
g.set(title='Tip vs Total Bill')
plt.show()
```

## 4. Categorical Plots

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 5))

sns.boxplot(data=tips, x='day', y='total_bill', hue='sex',
            palette='pastel', ax=axes[0])
axes[0].set_title('Box Plot')

sns.violinplot(data=tips, x='day', y='tip', inner='quartile',
               palette='Set2', ax=axes[1])
axes[1].set_title('Violin Plot')

sns.barplot(data=tips, x='day', y='total_bill', hue='sex',
            errorbar='sd', ax=axes[2])
axes[2].set_title('Bar Plot (mean + SD)')

plt.tight_layout()
plt.show()
```

## 5. Strip and Swarm Plots

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 5))

sns.stripplot(data=iris, x='species', y='petal_length',
              jitter=True, alpha=0.6, ax=axes[0])
axes[0].set_title('Strip Plot')

sns.swarmplot(data=iris, x='species', y='petal_length',
              size=4, ax=axes[1])
axes[1].set_title('Swarm Plot')

plt.tight_layout()
plt.show()
```

## 6. Heatmap

```python
corr = iris.drop(columns='species').corr().round(2)

fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(
    corr, annot=True, fmt='.2f',
    cmap='coolwarm', center=0,
    square=True, linewidths=0.5, ax=ax
)
ax.set_title('Iris Feature Correlations')
plt.tight_layout()
plt.show()
```

## 7. Pair Plot

```python
g = sns.pairplot(iris, hue='species', diag_kind='kde',
                 plot_kws={'alpha': 0.5})
g.figure.suptitle('Iris Pair Plot', y=1.02)
plt.show()
```

## 8. Joint Plot

```python
g = sns.jointplot(
    data=penguins,
    x='flipper_length_mm', y='body_mass_g',
    hue='species', height=6
)
g.set_axis_labels('Flipper Length (mm)', 'Body Mass (g)')
plt.suptitle('Penguins Joint Plot', y=1.01)
plt.show()
```

## 9. FacetGrid

```python
g = sns.FacetGrid(tips, col='time', row='sex', height=3.5, aspect=1.2)
g.map_dataframe(sns.histplot, x='total_bill', bins=15, kde=True)
g.add_legend()
g.figure.suptitle('Total Bill Distribution by Time and Sex', y=1.03)
plt.show()
```

## 10. Linear Regression Plot

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.regplot(data=tips, x='total_bill', y='tip',
            scatter_kws={'alpha': 0.4}, ax=axes[0])
axes[0].set_title('Linear Regression')

sns.residplot(data=tips, x='total_bill', y='tip',
              lowess=True, scatter_kws={'alpha': 0.4}, ax=axes[1])
axes[1].set_title('Residuals')

plt.tight_layout()
plt.show()
```

## 11. Theme Comparison

```python
themes = ['darkgrid', 'whitegrid', 'dark', 'white', 'ticks']
fig, axes = plt.subplots(1, len(themes), figsize=(16, 3))

for ax, theme in zip(axes, themes):
    with sns.axes_style(theme):
        temp_ax = fig.add_axes(ax.get_position())
        sns.lineplot(data=tips, x='size', y='total_bill',
                     estimator='mean', errorbar=None, ax=temp_ax)
        temp_ax.set_title(theme)
        ax.remove()

plt.suptitle('Seaborn Themes', y=1.06)
plt.show()
```
