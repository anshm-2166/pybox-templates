# Matplotlib - Plotting from First Principles
Line charts, scatter plots, histograms, subplots, 3D surfaces, and style customisation.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

print('Matplotlib version:', mpl.__version__)
plt.rcParams.update({'figure.dpi': 110, 'axes.spines.top': False,
                     'axes.spines.right': False})
```

## 1. Line Plot

```python
x = np.linspace(0, 2 * np.pi, 300)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, np.sin(x), label='sin(x)', lw=2)
ax.plot(x, np.cos(x), label='cos(x)', lw=2, linestyle='--')
ax.axhline(0, color='black', linewidth=0.5)
ax.set_title('Sine and Cosine')
ax.set_xlabel('x (radians)')
ax.set_ylabel('Amplitude')
ax.legend()
plt.tight_layout()
plt.show()
```

## 2. Scatter Plot with Color Mapping

```python
rng = np.random.default_rng(0)
n = 200
px = rng.standard_normal(n)
py = px * 0.6 + rng.standard_normal(n) * 0.8
colors = np.sqrt(px**2 + py**2)

fig, ax = plt.subplots(figsize=(6, 5))
sc = ax.scatter(px, py, c=colors, cmap='plasma', alpha=0.75, edgecolors='none')
fig.colorbar(sc, ax=ax, label='Distance from origin')
ax.set_title('Scatter - Correlated Variables')
ax.set_xlabel('X')
ax.set_ylabel('Y')
plt.tight_layout()
plt.show()
```

## 3. Bar Chart

```python
categories = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
values_a   = [45, 60, 55, 70, 65, 80]
values_b   = [30, 50, 48, 60, 55, 72]
x_pos = np.arange(len(categories))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(x_pos - width/2, values_a, width, label='Product A', color='steelblue')
ax.bar(x_pos + width/2, values_b, width, label='Product B', color='coral')
ax.set_xticks(x_pos)
ax.set_xticklabels(categories)
ax.set_title('Monthly Sales Comparison')
ax.set_ylabel('Units')
ax.legend()
plt.tight_layout()
plt.show()
```

## 4. Histogram

```python
rng2 = np.random.default_rng(1)
data1 = rng2.normal(0, 1, 1000)
data2 = rng2.normal(2, 1.5, 1000)

fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(data1, bins=40, alpha=0.6, label='N(0,1)', color='teal')
ax.hist(data2, bins=40, alpha=0.6, label='N(2,1.5)', color='tomato')
ax.set_title('Overlapping Histograms')
ax.set_xlabel('Value')
ax.set_ylabel('Frequency')
ax.legend()
plt.tight_layout()
plt.show()
```

## 5. Subplots Grid

```python
rng3 = np.random.default_rng(3)
t = np.linspace(0, 4 * np.pi, 400)

fig, axes = plt.subplots(2, 2, figsize=(10, 7))
fig.suptitle('Subplots Showcase', fontsize=14)

# Top-left - damped sine
axes[0, 0].plot(t, np.sin(t) * np.exp(-t / 8), color='steelblue')
axes[0, 0].set_title('Damped Sine')

# Top-right - step function
step_x = np.arange(10)
step_y = rng3.integers(1, 10, 10)
axes[0, 1].step(step_x, step_y, where='mid', color='darkorange')
axes[0, 1].set_title('Step Plot')

# Bottom-left - filled area
axes[1, 0].fill_between(t, np.sin(t), alpha=0.4, color='mediumseagreen')
axes[1, 0].plot(t, np.sin(t), color='darkgreen', lw=1)
axes[1, 0].set_title('Fill Between')

# Bottom-right - error bar
ex = np.arange(6)
ey = rng3.uniform(3, 8, 6)
err = rng3.uniform(0.3, 1.0, 6)
axes[1, 1].errorbar(ex, ey, yerr=err, fmt='o-', capsize=5, color='purple')
axes[1, 1].set_title('Error Bars')

plt.tight_layout()
plt.show()
```

## 6. Pie Chart

```python
labels  = ['Python', 'JavaScript', 'Rust', 'Go', 'Other']
sizes   = [38, 27, 12, 10, 13]
explode = (0.05, 0, 0, 0, 0)

fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(sizes, explode=explode, labels=labels, autopct='%1.1f%%',
       startangle=140, colors=plt.cm.Set2.colors)
ax.set_title('Language Popularity')
plt.tight_layout()
plt.show()
```

## 7. Heatmap with imshow

```python
corr_data = np.corrcoef(np.random.default_rng(5).standard_normal((6, 100)))
labels_h = ['A', 'B', 'C', 'D', 'E', 'F']

fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(corr_data, cmap='coolwarm', vmin=-1, vmax=1)
fig.colorbar(im, ax=ax)
ax.set_xticks(range(6)); ax.set_xticklabels(labels_h)
ax.set_yticks(range(6)); ax.set_yticklabels(labels_h)
ax.set_title('Correlation Heatmap')

for i in range(6):
    for j in range(6):
        ax.text(j, i, f'{corr_data[i, j]:.2f}', ha='center', va='center',
                fontsize=8, color='black')

plt.tight_layout()
plt.show()
```

## 8. 3D Surface Plot

```python
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

x3 = np.linspace(-3, 3, 60)
y3 = np.linspace(-3, 3, 60)
X3, Y3 = np.meshgrid(x3, y3)
Z3 = np.sin(np.sqrt(X3**2 + Y3**2))

fig = plt.figure(figsize=(8, 6))
ax3 = fig.add_subplot(111, projection='3d')
surf = ax3.plot_surface(X3, Y3, Z3, cmap='viridis', linewidth=0, antialiased=True)
fig.colorbar(surf, ax=ax3, shrink=0.5)
ax3.set_title('3D Surface - sin(r)')
plt.tight_layout()
plt.show()
```

## 9. Annotations and Custom Styles

```python
x_ann = np.linspace(0, 2 * np.pi, 200)
y_ann = np.sin(x_ann)
peak_x = x_ann[np.argmax(y_ann)]
peak_y = y_ann.max()

with plt.style.context('seaborn-v0_8-whitegrid'):
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.plot(x_ann, y_ann, color='navy', lw=2)
    ax.annotate(
        f'Peak ({peak_x:.2f}, {peak_y:.2f})',
        xy=(peak_x, peak_y),
        xytext=(peak_x + 0.8, peak_y - 0.2),
        arrowprops=dict(arrowstyle='->', color='crimson'),
        fontsize=10, color='crimson'
    )
    ax.set_title('Annotated Plot')
    plt.tight_layout()
    plt.show()
```

## 10. Saving Figures

```python
fig_save, ax_save = plt.subplots(figsize=(6, 3))
ax_save.plot(np.linspace(0, 10, 100), np.exp(-np.linspace(0, 10, 100) / 3),
             color='darkcyan', lw=2)
ax_save.set_title('Exponential Decay')
ax_save.set_xlabel('t')
ax_save.set_ylabel('f(t)')

fig_save.savefig('output_plot.png', dpi=150, bbox_inches='tight')
print('Figure saved to output_plot.png')
plt.show()
```
